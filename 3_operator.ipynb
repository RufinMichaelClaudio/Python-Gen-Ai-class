```python
# Comparison operators 
# ==, !=, <, >, <=, >=

print(10 == 10)
print(10 != 10)
print(10 < 10)
print(10 <= 10)
print(10 > 10)
print(10 >= 10)



```

```python
# Logical operators 
# and, or, not 

print('and')

print(10 == 10 and 6 > 4)
print(10 != 10 and 6 > 4)
print(10 == 10 and 6 < 4)
print(10 != 10 and 6 < 4)

print
```
