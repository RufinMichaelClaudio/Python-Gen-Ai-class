```python
# Comparison operators 
# ==, !=, <, >, <=, >=

print(10 == 10)
print(10 != 10)
print(10 < 10)
print(10 <= 10)
print(10 > 10)
print(10 >= 10)



```

```python
# Logical operators 
# and, or, not 

print('and')

print(10 == 10 and 6 > 4)
print(10 != 10 and 6 > 4)
print(10 == 10 and 6 < 4)
print(10 != 10 and 6 < 4)

print('or')
print(10 == 10 or 6 > 4)
print(10 != 10 or 6 > 4)
print(10 == 10 or 6 < 4)
print(10 != 10 or 6 < 4)

print('not')
print(not 10 == 10)
print(not 10 != 10)
```
