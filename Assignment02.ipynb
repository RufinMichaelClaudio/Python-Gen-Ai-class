```python
print(10+3)
print(10-3)
print(10*3)
print(10/3)
print(10**3)
print(10//3)
print(10%3)
```

```python
print(10==10)
print(10!=10)
print(10>10)
print(10<10)
print(10<=10)
```

```python
print('----AND----')

print(10==10 and 6<4)
print(10!=10 and 6<4)
print(10==10 and 6>4)
print(10!=10 and 6>4)

print('----OR----')
print(10==10 or 6>4)
print(10!=10 or 6>4)
print(10==10 or 6<4)
print(10!=10 or 6<4)

print('----Not----')
print(not 10==10)
print(not 10!=10)
```

```python
num=int(input("enter a number"))
if num>10:
    print("the number is postive")
elif num<10:
    print("the number is negative")
else:
    print("print the number is zero")
```

```python
num=int(input("enter a number"))
if num % 2==0:
    print("the number is even")
else:
    print("the number is odd")
```
