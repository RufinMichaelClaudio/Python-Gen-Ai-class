```python
mystr = "racecar"
revstr = mystr[::-1]

print(mystr)
print(revstr)

if mystr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "MaLaYa LAm"
newstr = mystr.lower().replace(" ",' ')
revstr = newstr[::-1]

print(mystr)
print(newstr)
print(revstr)

if newstr == revstr:
    print("true")
else:
    print("false")
```

```python
mystr = "Sir, I demand, I am a maid named Iris"
newstr = mystr.lower().replace(" ",'').replace(',','')
revstr = newstr[::-1]

print(mystr)
print(newstr)
print(revstr)

if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "Red roses run no risk. Sir, on Nurse's Order. "
newstr = ''
for each in mystr:
    if each.isalpha():
        newstr += each.lower() # newstr += each.lower()
print(newstr)
if newstr == newstr[::-1]:
    print(True)
else:
    print(False)
```
