Define a class Car with attributes brand and model. Create an object of the class and display its 
details.

```python
class Car:
    def __init__ (self, brand, model):
        print("Car Data")
        self.brand = brand
        self.model = model

    def display (self):
        print(f"the car brand is = {self.brand}")
        print(f"the car model is = {self.model}")

obj1 = Car("toyota","Corolla")
obj1.display()
print('----------------')

obj2 = Car("Honda","City")
obj2.display()
```

Write a class Student with attributes name and marks. Add a method display() to print details. 

```python
class Student:

    def __init__(self, name,marks):
        print("Student Data")
        self.name = name 
        self.marks = marks

    def display(self):
        print(f"The student name is = {self.name}")
        print(f"The student marks are = {self.marks}")

obj1 = Student("rufin",98)
obj1.display()

print('-----------------')

obj2 = Student("michael",97)
obj2.display()
```

create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.

```python
class Rectangle:

    def __init__(self,length,width):
        print("Rectangle Data")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

obj1 = Rectangle(10,5)
print(f"The area is = {obj1.area()}")
print(f"The perimeter is = {obj1.perimeter()}")

print('-------------------')

print(f"The area is = {obj2.area()}")
print(f"The perimeter is = {obj2.perimeter()}")

```
