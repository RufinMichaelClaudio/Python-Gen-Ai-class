create a class Rectangle with attributes length and width. Add methods to calculate area and perimeter.

```python
class Rectangle:

    def __init__(self,length,width):
        print("Rectangle Data")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

obj1 = Rectangle(10,5)
print(f"The area is = {obj1.area()}")
print(f"The perimeter is = {obj1.perimeter()}")

print('-------------------')

obj2 = Rectangle(7,3)
print(f"The area is = {obj2.area()}")
print(f"The perimeter is = {obj2.perimeter()}")

```

Create a class BankAccount with attributes account_number, name, and balance. Add methods to deposit, withdraw, and display balance.

```python
class BankAccount:

    def __init__(self,account_number,name,balance=0):
        print("account created")
        self.account_number = account_number
        self.name = name
        self.balance = balance 

    def deposit(self,amount):
        self.balance = self.balance + amount 
        print(f"rs.{amount} deposited successfully.")

    def withdraw(self,amount):
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance = self.balance - amount 
            print(f"Rs.{amount} withdrawn successfully.")

    def display_balance(self):
        print(f"account number : {self.account_number}")
        print(f"name           : {self.name}")
        print(f"balance        : rs.{self.balance}")

acc1 = BankAccount(1001,'rufin',5000)
acc1.display_balance()

print('------------------------------')

acc1.deposit(20000)
acc1.withdraw(300000)
acc1.withdraw(30000)

print('-----------------')

acc1.display_balance()
```

Write a class Employee with attributes emp_id, name, and salary. Add a method to give a 10% bonus to employees. 

```python
class employee:

    def __init__(self,emp_id,name,salary):
        print("employee data")
        self.emp_id = emp_id
        self.name = name 
        self.salary = salary

    def give_bonus(self):
        bonus = self.salary *10 / 100
        self.salary = self.salary + bonus 
        print(f"bonus of rs.{bonus} added to {self.name}.")

    def display(self):
        print(f"employee id : {self.emp_id}")
        print(f"name        : {self.name}")
        print(f"salary      : rs.{self.salary}")

emp1 = employee(101,"arun",50000)
emp1.display()

print('----------------------------')

emp1.give_bonus()

print('-------------------')

emp1.display()
```

Implement a class Library with methods to add_book(title), remove_book(title), and display_books().


```python
class library: 

    def __init__(self):
        print("library created")
        self.books = []

    def add_books(self,title):
        self.books.append(title)
        print(f"'{title}'added.")

    def remove_book(self,title):
        if title in self.books:
            self.books.remove(title)
            print(f"'{title}' removed.")
        else:
            print(f"'{title}' not found in the libuary.")

    def display_books(self):
        if len(self.books) == 0:
            print("no books in the library.")
        else:
            print("books in the library:")
            count = 1
            for book in self.books:
                print(f"{count}. {book}")
                count = count + 1

lib = library()
lib.display_books()

print('-------------------')

lib.add_books("python basics")
lib.add_books("data science")
lib.add_books("biotechnology")

print('---------------------')

lib.display_books()

print('--------------------------')

lib.remove_book("data science")
lib.remove_book("physics")

print('-----------------------')

lib.display_books()


```

Create a class Calculator with methods add, subtract, multiply, and divide. Demonstrate with objects.

```python
class calculator:

    def __init__(self):
        print("calculator ready")

    def add(self,a,b):
        return a + b 

    def subtract(self,a,b):
        return a - b 

    def multiply(self,a,b):
        return a * b 

    def divide(self,a,b):
        if b == 0:
            return "Error: Cannot divide by zero!"
        else:
            return a / b 

calc1 = calculator()
print(f"addition       : {calc1.add(10,5)}")
print(f"subtraction    : {calc1.subtract(10,5)}")
print(f"multiplication : {calc1.multiply(10,5)}")
print(f"division       : {calc1.divide(10,5)} ")
print(f"divide by zero : {calc1.divide(10,0)}")

print('-----------------------')

calc2 = calculator()
print(f"addiction      : {calc2.add(7.5,2.5)}")
print(f"subtraction    : {calc2.subtract(7.5,2.5)}")
print(f"multiplication : {calc2.multiply(7.5,2.5)}")
print(f"division       : {calc2.divide(7.5,2.5)}")
```
