```python
class print(ABC):
    @abstractmethod
    def format_message(self, *values, sep=" "):
        """Convert arguments into a single printable string."""
        return sep.join(str(v) for v in values)

    @abstractmethod
    def write(self, text, file=None):
        """Write text to the provided output stream."""
        if file is None:
            file = sys.stdout
        file.write(text)

    def __call__(self, *values, sep=" ", end="\n", file=None, flush=False):
        text = self.format_message(*values, sep=sep) + end
        self.write(text, file=file)
```
