```python
score = 100
secret_number = 50
guess = int(input("enter your guess:    "))

if guess == secret_number:
    print("correct!")
    score = score + 20
    print("your Score:", score)

elif guess > secret_number:
    print("Too high!")
    score = score -10
    print("your score:", score)

else:
    print("Too Low!")
    score = score - 10
    print("your score:", score)
```
