```python
# keys ()

mydict ={
    "name":"rufin",
    "age" : 22,
    "city":"chennai",
      500 : 50.55,
}
print(mydict)
print()
print(list(mydict.keys()))
print()

ans = list(mydict.keys())
print(ans[1])

print()
for each in mydict.keys():
    print(each)
```

```python
#values 

mydict = {
    "name":"rufin",
    "age" :22,
    "city":"madurai",
      500 : 50.55,   
}
print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```
